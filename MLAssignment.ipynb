#Retail Market Basket Analysis & Sales Forecasting

#
#  **Project Objective:**
# The goal of this project is to analyze a large transactional dataset from a UK-based online retailer. We will perform a comprehensive Exploratory Data Analysis (EDA) and apply two distinct machine learning techniques:
# 1.  **Unsupervised Learning:** Using the **Apriori algorithm** for Market Basket Analysis to discover which products are frequently purchased together.
# 2.  **Supervised Learning:** Building a **SARIMA time-series model** to forecast future sales revenue.
#
# The final output will be a complete analysis notebook with actionable insights for marketing, inventory management, and strategic planning.
#
# ---        
#
#  **Phase 1: Data Loading and Preprocessing**
#
# In this initial phase, we will load the dataset and perform essential cleaning operations. This is the most critical step, as the quality of our analysis depends entirely on the quality of our data.

#  1.1: Import Libraries and Load Data
# First, we import all the necessary Python libraries for data manipulation, visualization, and machine learning. We then use Google Colab's built-in file uploader to load the dataset.

```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from mlxtend.frequent_patterns import apriori, association_rules
from mlxtend.preprocessing import TransactionEncoder
from google.colab import files
import gc
from statsmodels.tsa.seasonal import seasonal_decompose
from statsmodels.tsa.statespace.sarimax import SARIMAX
import warnings
warnings.filterwarnings("ignore")
sns.set(style="whitegrid", palette="muted")
plt.rcParams['figure.figsize'] = (10, 5)

print("--- Market Analysis Notebook Started ---")
print("\n[Phase 1] Loading and cleaning data...")

# Use Colab's file uploader
print("Please upload your Excel data file (.xlsx)")
uploaded = files.upload()

if not uploaded:
    print("\nNo file was uploaded. Halting the script.")
    exit()

file_name = list(uploaded.keys())[0]
print(f"\nUser uploaded file: '{file_name}'")

try:
    df = pd.read_excel(file_name)
    print(f"Successfully loaded the dataset. Initial shape: {df.shape}")
except Exception as e:
    print(f"\nError reading the Excel file: {e}")
    exit()


# 1.2: Data Cleaning
# Now, we proceed with the cleaning steps to ensure data integrity.

# Drop CustomerID
df.drop('CustomerID', axis=1, inplace=True, errors='ignore')

# Handle missing Itemname
df.dropna(axis=0, subset=['Itemname'], inplace=True)

# Clean up Itemname text
df['Itemname'] = df['Itemname'].str.strip()

# Remove cancelled orders
df = df[~df['BillNo'].astype(str).str.startswith('C')]

# Ensure Quantity is positive
df = df[df['Quantity'] > 0]

# Convert data types
df['Date'] = pd.to_datetime(df['Date'])
df['Quantity'] = pd.to_numeric(df['Quantity'], downcast='integer')
df['Price'] = pd.to_numeric(df['Price'], downcast='float')

# Create a 'Revenue' column
df['Revenue'] = df['Quantity'] * df['Price']

print(f"Data cleaning complete. Final shape of the dataset: {df.shape}")
print("\n--- First 5 Rows of Cleaned Data ---")
display(df.head())

# ## **Phase 2: Exploratory Data Analysis (EDA)**
# Now that our data is clean, we can explore it to find trends and patterns.
# Let's start with fundamental business metrics.

total_revenue = df['Revenue'].sum()
total_orders = df['BillNo'].nunique()
total_items_sold = df['Quantity'].sum()
total_unique_products = df['Itemname'].nunique()

print("--- Key Business Metrics ---")
print(f"Total Revenue:         ${total_revenue:,.2f}")
print(f"Total Unique Orders:    {total_orders:,}")
print(f"Total Items Sold:       {total_items_sold:,}")
print(f"Total Unique Products:  {total_unique_products:,}")
print("----------------------------")

# 2.2a: Global Sales Trend Over Time
# This visualization helps us understand the business's growth and identify any seasonal patterns in sales.

df['Month'] = df['Date'].dt.to_period('M')
monthly_sales = df.groupby('Month')['Revenue'].sum().reset_index()
monthly_sales['Month'] = monthly_sales['Month'].astype(str)

plt.figure(figsize=(14, 6))
sns.lineplot(x='Month', y='Revenue', data=monthly_sales, marker='o', color='royalblue')
plt.title('Total Monthly Sales Revenue Over Time', fontsize=16, fontweight='bold')
plt.xlabel('Month', fontsize=12, fontweight='bold')
plt.ylabel('Total Revenue ($)', fontsize=12, fontweight='bold')
plt.xticks(rotation=45, ha='right')
sns.despine()
plt.tight_layout()
plt.show()

# 2.2b: Month-over-Month Sales Comparison by Year
# This bar chart provides a more granular view of seasonality by comparing sales for the same month across different years. This helps to confirm if the end-of-year sales peak is a consistent pattern.

df['Year'] = df['Date'].dt.year
df['MonthName'] = df['Date'].dt.month_name()

# Order the months correctly for plotting
month_order = ['January', 'February', 'March', 'April', 'May', 'June', 'July', 'August', 'September', 'October', 'November', 'December']

monthly_comparison = df.groupby(['Year', 'MonthName'])['Revenue'].sum().reset_index()

plt.figure(figsize=(14, 7))
sns.barplot(x='MonthName', y='Revenue', hue='Year', data=monthly_comparison, order=month_order, palette='viridis')
plt.title('Month-over-Month Sales Comparison by Year', fontsize=16, fontweight='bold')
plt.xlabel('Month', fontsize=12, fontweight='bold')
plt.ylabel('Total Revenue ($)', fontsize=12, fontweight='bold')
plt.xticks(rotation=45, ha='right')
plt.legend(title='Year')
sns.despine()
plt.tight_layout()
plt.show()


# 2.3: Top 20 Best-Selling Products (Global)
# This chart shows the most popular products, crucial for inventory management.

top_products = df.groupby('Itemname')['Quantity'].sum().sort_values(ascending=False).head(20)

plt.figure(figsize=(10, 8))
sns.barplot(x=top_products.values, y=top_products.index, palette='magma')
plt.title('Top 20 Best-Selling Products (by Quantity)', fontsize=16, fontweight='bold')
plt.xlabel('Total Quantity Sold', fontsize=12, fontweight='bold')
plt.ylabel('Product Name', fontsize=12, fontweight='bold')
sns.despine()
plt.show()


# 2.4: Top 10 Countries by Revenue
# Understanding our customer geography is key to tailoring marketing efforts.

top_countries = df.groupby('Country')['Revenue'].sum().sort_values(ascending=False).head(10)

plt.figure(figsize=(10, 6))
sns.barplot(x=top_countries.values, y=top_countries.index, palette='plasma')
plt.title('Top 10 Countries by Total Sales Revenue', fontsize=16, fontweight='bold')
plt.xlabel('Total Revenue ($)', fontsize=12, fontweight='bold')
plt.ylabel('Country', fontsize=12, fontweight='bold')
sns.despine()
plt.show()

# 2.5: Sales by Time of Day and Day of Week
# This analysis helps identify peak shopping times, which is highly actionable for staffing, server load management, and targeted promotions.

df['Hour'] = df['Date'].dt.hour
df['DayOfWeek'] = df['Date'].dt.day_name()

# Sales by Hour
hourly_sales = df.groupby('Hour')['Revenue'].sum()

# Sales by Day of Week
weekly_sales = df.groupby('DayOfWeek')['Revenue'].sum().reindex(
    ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']
)

fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(12, 10))

# Hourly Sales Plot
sns.barplot(x=hourly_sales.index, y=hourly_sales.values, ax=ax1, palette='coolwarm')
ax1.set_title('Sales Revenue by Hour of the Day', fontsize=14, fontweight='bold')
ax1.set_xlabel('Hour of Day (24-hour format)', fontsize=12, fontweight='bold')
ax1.set_ylabel('Total Revenue ($)', fontsize=12, fontweight='bold')

# Weekly Sales Plot
sns.barplot(x=weekly_sales.index, y=weekly_sales.values, ax=ax2, palette='winter')
ax2.set_title('Sales Revenue by Day of the Week', fontsize=14, fontweight='bold')
ax2.set_xlabel('Day of the Week', fontsize=12, fontweight='bold')
ax2.set_ylabel('Total Revenue ($)', fontsize=12, fontweight='bold')

sns.despine()
plt.tight_layout()
plt.show()

# 2.6: Distribution of Order Values
# This visualization shows how much customers typically spend per transaction. We can see if the business relies on many small purchases or fewer large ones.

order_value = df.groupby('BillNo')['Revenue'].sum()

plt.figure(figsize=(12, 6))
sns.histplot(order_value, bins=50, kde=True, color='purple')
plt.title('Distribution of Order Value (Revenue per Transaction)', fontsize=16, fontweight='bold')
plt.xlabel('Order Value ($)', fontsize=12, fontweight='bold')
plt.ylabel('Number of Orders', fontsize=12, fontweight='bold')
plt.xlim(0, 2000) # Limit x-axis to see the distribution of typical orders
plt.axvline(order_value.median(), color='red', linestyle='--', label=f'Median Order Value: ${order_value.median():.2f}')
plt.legend()
sns.despine()
plt.show()

#  **Phase 3: Deep Dive into Country-Level Analysis**
# Global trends are useful, but performance can vary significantly by region.
#3.1: Country Performance Summary
# "Average Revenue Per Order" is a great metric for understanding the purchasing power of customers in different regions.

country_agg = df.groupby('Country').agg(
    TotalRevenue=('Revenue', 'sum'),
    TotalOrders=('BillNo', 'nunique')
).sort_values(by='TotalRevenue', ascending=False)
country_agg['AvgRevenuePerOrder'] = country_agg['TotalRevenue'] / country_agg['TotalOrders']

print("\n--- Country Performance Summary (Top 10) ---")
display(country_agg.head(10).style.set_caption("Financial Overview of Top 10 Countries").format({
    'TotalRevenue': '${:,.2f}',
    'TotalOrders': '{:,}',
    'AvgRevenuePerOrder': '${:,.2f}'
}))

# ## **Phase 4: Market Basket Analysis (Unsupervised Machine Learning)**
# This technique helps us answer: **"If a customer buys product A, how likely are they to also buy product B?"**
# 4.1: Data Preparation for the Apriori Algorithm
# To make the analysis faster and more relevant, we will focus only on the largest market, the **United Kingdom**.

print("\n[Phase 4] Performing Market Basket Analysis...")
print("This step might take several minutes...")

df_uk = df[df['Country'] == 'United Kingdom']

# Create a list of lists, where each inner list is a transaction
transactions = df_uk.groupby('BillNo')['Itemname'].apply(list).tolist()

# Use the memory-efficient TransactionEncoder
te = TransactionEncoder()
te_ary = te.fit(transactions).transform(transactions)
basket_sets = pd.DataFrame(te_ary, columns=te.columns_)
print("- Transaction matrix created for Apriori algorithm.")

# Free up memory
del transactions, df_uk
gc.collect()

# 4.2: Running the Algorithm and Generating Rules
# Now we run the Apriori algorithm and generate association rules. We keep only rules with a **lift > 1**, as these represent meaningful positive associations.

frequent_itemsets = apriori(basket_sets, min_support=0.02, use_colnames=True)
rules = association_rules(frequent_itemsets, metric="lift", min_threshold=1)
print("- Association rules generated successfully.")


# 4.3: Analyzing the Results
# The table below shows the top 20 strongest product associations, sorted by **lift**. This is the direct, actionable output of our machine learning model.

rules['antecedents'] = rules['antecedents'].apply(lambda x: ', '.join(list(x)))
rules['consequents'] = rules['consequents'].apply(lambda x: ', '.join(list(x)))
rules = rules.sort_values(by='lift', ascending=False)

print("\n--- Top 20 Product Association Rules (Sorted by Lift) ---")
display(rules.head(20).style.set_caption("Strongest Product Affinities from Market Basket Analysis").format({
    'support': '{:.3f}',
    'confidence': '{:.3f}',
    'lift': '{:.2f}'
}))

# Phase 5: Sales Forecasting (Supervised Machine Learning)
# We will now build a time-series model to predict future sales. This is a supervised learning task where we use historical data to forecast the total revenue for the upcoming month.
# 5.1: Preparing Data for Time-Series Analysis
# Forecasting models require the data to be in a time-series format. We will aggregate our total revenue for each day.

daily_sales = df.set_index('Date').resample('D')['Revenue'].sum().reset_index()
daily_sales = daily_sales.set_index('Date')


# 5.2: Decomposing the Time Series
decomposition = seasonal_decompose(daily_sales['Revenue'], model='additive', period=7)

fig, (ax1, ax2, ax3, ax4) = plt.subplots(4, 1, figsize=(12, 10), sharex=True)
decomposition.observed.plot(ax=ax1, legend=False, color='dodgerblue')
ax1.set_ylabel('Observed', fontweight='bold')
decomposition.trend.plot(ax=ax2, legend=False, color='dodgerblue')
ax2.set_ylabel('Trend', fontweight='bold')
decomposition.seasonal.plot(ax=ax3, legend=False, color='dodgerblue')
ax3.set_ylabel('Seasonal', fontweight='bold')
decomposition.resid.plot(ax=ax4, legend=False, color='dodgerblue', marker='o', linestyle='None', markersize=2)
ax4.set_ylabel('Residual', fontweight='bold')
plt.suptitle('Time Series Decomposition of Daily Sales', fontsize=18, fontweight='bold')
sns.despine()
plt.tight_layout(rect=[0, 0.03, 1, 0.97])
plt.show()


# 5.3: Building and Training the SARIMA Model
# Given the clear weekly seasonality, a **SARIMA (Seasonal AutoRegressive Integrated Moving Average)** model is an excellent choice. We will train the model on all data except the last 30 days, which we'll use as a "test set" to validate our predictions.

# Split data into training and test sets
train_data = daily_sales[:-30]
test_data = daily_sales[-30:]

print(f"Training data from {train_data.index.min()} to {train_data.index.max()}")
print(f"Test data from {test_data.index.min()} to {test_data.index.max()}")
model = SARIMAX(train_data['Revenue'],
                order=(1, 1, 1),
                seasonal_order=(1, 1, 1, 7))

# Fit the model
print("\nTraining the SARIMA model... This may take a moment.")
results = model.fit(disp=False)
print("Model training complete.")


# 5.4: Visualizing Forecast vs. Actual Sales
# Now, let's see how well our model did. We'll ask it to predict the sales for the 30-day test period and plot those predictions against the actual sales data we held out.

# Generate predictions for the test period
predictions = results.get_prediction(start=test_data.index[0], end=test_data.index[-1])
predicted_mean = predictions.predicted_mean

# Plotting the results
plt.figure(figsize=(14, 7))
plt.plot(daily_sales.index, daily_sales['Revenue'], label='Historical Daily Sales', alpha=0.7)
plt.plot(predicted_mean.index, predicted_mean.values, color='red', label='Predicted Sales (Test Set)')
plt.title('Sales Forecast vs. Actual Sales (Last 30 Days)', fontsize=16, fontweight='bold')
plt.xlabel('Date', fontsize=12, fontweight='bold')
plt.ylabel('Revenue ($)', fontsize=12, fontweight='bold')
plt.legend()
sns.despine()
plt.show()


#5.5: Forecasting Future Sales
# Finally, we can use our model to forecast sales for the next 30 days. We first retrain the model on the *entire* dataset to give it the most information possible, then ask it to predict the future.

print("\nRetraining model on all data to forecast the future...")
full_model = SARIMAX(daily_sales['Revenue'],
                     order=(1, 1, 1),
                     seasonal_order=(1, 1, 1, 7))
full_results = full_model.fit(disp=False)

# Forecast the next 30 days
forecast = full_results.get_forecast(steps=30)
forecast_mean = forecast.predicted_mean
forecast_ci = forecast.conf_int() # Confidence intervals

# Plotting the future forecast
plt.figure(figsize=(14, 7))
plt.plot(daily_sales.index, daily_sales['Revenue'], label='Historical Daily Sales')
plt.plot(forecast_mean.index, forecast_mean.values, color='red', linestyle='--', label='Future Forecast')
plt.fill_between(forecast_ci.index,
                 forecast_ci.iloc[:, 0],
                 forecast_ci.iloc[:, 1], color='pink', alpha=0.5, label='95% Confidence Interval')
plt.title('Sales Forecast for the Next 30 Days', fontsize=16, fontweight='bold')
plt.xlabel('Date', fontsize=12, fontweight='bold')
plt.ylabel('Revenue ($)', fontsize=12, fontweight='bold')
plt.legend()
sns.despine()
plt.show()


try:
    first_rule = rules.iloc[0]
    rule_antecedent = first_rule['antecedents']
    rule_consequent = first_rule['consequents']
    rule_lift = first_rule['lift']
    print("\n[Dynamic Conclusion Text Generated]")
except (IndexError, NameError):
    print("\nCould not generate dynamic conclusion text as no rules were found.")
```
